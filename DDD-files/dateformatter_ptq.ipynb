# Dateformatter

#### Description
 
This notebook transforms the first column of the ptq files into a format the DDD model can read. The datetime column is changed into the format year-month-day-hour-minute. Example for May 1st 2009 18:00 --> 2009-5-1-18-00 in each separate column (5 columns).

```python
# USE THIS FOR ALL HSO GENERATED PTQS, DIFFERENT COLUMN HEADERS
def dateformatter(file_path):
    data = pd.read_csv(file_path, sep=";")
    data.rename(columns={'Tidspunkt': 'datetime'}, inplace=True) # For Nigardsbrevatnet
    data["datetime"] = pd.to_datetime(data["datetime"])
    data["year"] = data["datetime"].dt.year
    data["month"] = data["datetime"].dt.month
    data["day"] = data["datetime"].dt.day
    data["hour"] = data["datetime"].dt.hour
    data["minute"] = data["datetime"].dt.minute

    df = data.drop(columns=["datetime"])
    cols = ["year", "month", "day", "hour", "minute"]
    df = df[cols + [c for c in df.columns if c not in cols]]
    df["q"]=1 # for all catchments without observed discharge
    #df["q"]=df["q"]#*0.282 # for scaling observed discharge by area factor, for Båtedøla S 
    if catchment == "Gaupne":
        df = df.drop(df.index[105758:105762])
        df.to_csv(f"{catchment}_1H_ptq_edited.csv", sep=";", header=False, index=False)
    else:
        df.to_csv(f"{catchment}_1H_ptq_edited.csv", sep=";", header=False, index=False)
    return df

file_path = f"{catchment}_1H_ptq.csv"
dateformatter(file_path)
```

# Editing the harddrive forecast files 

```python
import pandas as pd
import glob
import os

# -------------------------
# Folder
# -------------------------
# folder = r"D:\Master_project-main\Master_project-main\EC_forecasts\ptq_hourly_ECENS" # for ec ens
# folder = r"D:\Master_project-main\Master_project-main\EC_forecasts\ptq_hourly_HRES" # for ec hres
folder = r"D:\Master_project-main\Master_project-main\EC_forecasts\ptq_hourly_MEPS" # for MEPS

# pattern = os.path.join(folder, "Styggevatn_EC_ENS_2018*T00_member_*.csv") # for ec ens
# pattern = os.path.join(folder, "Styggevatn_EC_HRES_2018*T*_ptq_hourly.csv") # for ec hres
pattern = os.path.join(folder, "Styggevatn_MEPS_2018*T*_ptq_hourly.csv") # for meps

# pattern = os.path.join(folder, "Tunsbergdalsvatn_EC_ENS_2018*T00_member_*.csv") # for ec ens
# pattern = os.path.join(folder, "Tunsbergdalsvatn_EC_HRES_2018*T*_ptq_hourly.csv") # for ec hres
# pattern = os.path.join(folder, "Tunsbergdalsvatn_MEPS_2018*T*_ptq_hourly.csv") # for meps

# pattern = os.path.join(folder, "Gaupne_EC_ENS_2018*T00_member_*.csv") # for ec ens
# pattern = os.path.join(folder, "Gaupne_EC_HRES_2018*T*_ptq_hourly.csv") # for ec hres
# pattern = os.path.join(folder, "Gaupne_MEPS_2018*T*_ptq_hourly.csv") # for meps

files = glob.glob(pattern)

print(f"Found {len(files)} files")

# -------------------------
# Function
# -------------------------
def dateformatter(file_path):
    #data = pd.read_csv(file_path, sep=";")
    data = pd.read_csv(file_path, sep=",") # My ec hres and meps are comma seperated for some reason...

    # Detect datetime column
    if "datetime" in data.columns:
        data["datetime"] = pd.to_datetime(data["datetime"])
        date_col = "datetime"
    elif "date" in data.columns:
        data["date"] = pd.to_datetime(data["date"])
        date_col = "date"
    elif "time" in data.columns:
        data["time"] = pd.to_datetime(data["time"])
        date_col = "time"   
    elif "index" in data.columns:
        data["index"] = pd.to_datetime(data["index"])
        date_col = "index"  
    else:
        raise ValueError(f"No datetime column in {file_path}")

    # Extract components
    data["year"] = data[date_col].dt.year
    data["month"] = data[date_col].dt.month
    data["day"] = data[date_col].dt.day
    data["hour"] = data[date_col].dt.hour
    data["minute"] = data[date_col].dt.minute
    

    # Drop original datetime column
    df = data.drop(columns=[date_col])

    # Reorder columns (DDD format)
    cols = ["year", "month", "day", "hour", "minute"]
    df = df[cols + [c for c in df.columns if c not in cols]]
    #df["q"]=1
    # Overwrite WITHOUT headers
    df.to_csv(file_path, sep=";", header=False, index=False)

    print(f"✔ Done: {os.path.basename(file_path)}")


# -------------------------
# LOOP
# -------------------------
for file in files:
    try:
        dateformatter(file)
    except Exception as e:
        print(f"❌ Error in {file}: {e}")

print("All files formatted for DDD ✅")
```

# Recovering the datetime column

```python
import pandas as pd
import glob
import os

# folder = r"D:\Master_project-main\Master_project-main\EC_forecasts\ptq_hourly_HRES"
folder = r"D:\Master_project-main\Master_project-main\EC_forecasts\ptq_hourly_ECENS"
# pattern = os.path.join(folder, "Gaupne_EC_ENS_2018*T*_ptq_hourly.csv")
pattern = os.path.join(folder, "Tunsbergdalsvatn_EC_ENS_2018*T*_ptq_hourly.csv")
files = glob.glob(pattern)

print(f"Found {len(files)} files")

def restore_datetime(file_path):
    # Read WITHOUT header
    df = pd.read_csv(file_path, sep=";", header=None)

    # Recreate column names
    base_cols = ["year", "month", "day", "hour", "minute"]
    other_cols = [f"col_{i}" for i in range(len(df.columns) - 5)]
    df.columns = base_cols + other_cols

    # Reconstruct datetime
    df["datetime"] = pd.to_datetime(
        df[["year", "month", "day", "hour", "minute"]]
    )

    # Drop split columns
    df = df.drop(columns=["year", "month", "day", "hour", "minute"])

    # Move datetime to front
    cols = ["datetime"] + [c for c in df.columns if c != "datetime"]
    df = df[cols]

    # Save (now WITH header and comma-separated like original)
    df.to_csv(file_path, index=False)

    print(f"✔ Restored: {os.path.basename(file_path)}")


for file in files:
    try:
        restore_datetime(file)
    except Exception as e:
        print(f"❌ Error in {file}: {e}")

print("All files restored ✅")
```

## Merging EC HRES and observed from 2016

```python
import pandas as pd
import glob
import os

# -------------------------
# Paths
# -------------------------

# obs_path="Tunsbergdalsvatnet_1H_ptq.csv"
# obs_path="Gaupne_1H_ptq.csv"
obs_path="Austdalsvatnet_1H_ptq.csv"


# folder = r"D:\Master_project-main\Master_project-main\EC_forecasts\ptq_hourly_HRES"
# pattern = os.path.join(folder, "Tunsbergdalsvatn_EC_HRES_2018*T*_ptq_hourly.csv")

# folder = r"D:\Master_project-main\Master_project-main\EC_forecasts\ptq_hourly_ECENS"
# pattern = os.path.join(folder, "Tunsbergdalsvatn_EC_ENS_2018*T00_member_*_ptq_hourly.csv")

folder = r"D:\Master_project-main\Master_project-main\EC_forecasts\ptq_hourly_MEPS"
pattern = os.path.join(folder, "Styggevatn_MEPS_2018*T*_ptq_hourly.csv")

files = glob.glob(pattern)

print(f"Found {len(files)} forecast files")
# --- Load observed data ---
obs = pd.read_csv(obs_path, sep=";")
obs["datetime"] = pd.to_datetime(obs["datetime"])
obs.set_index("datetime", inplace=True)

# Crop start (spin-up)
obs = obs.loc["2016-01-01":]

for file in files:
    try:
        # --- Load forecast ---
        fc = pd.read_csv(file)
        fc = fc.rename(columns={"time": "datetime"}) # if column name is time, not datetime (check)
        fc["datetime"] = pd.to_datetime(fc["datetime"])
        fc.set_index("datetime", inplace=True)

        # --- Rename forecast columns to match obs ---
        rename_dict = {}

        # Precip
        for i in range(10):
            rename_dict[f"P_band_{i+1}"] = f"rr{i}"

        # Temp
        for i in range(10):
            rename_dict[f"T_band_{i+1}"] = f"tm{i}"

        fc = fc.rename(columns=rename_dict)

        # --- Get forecast start ---
        fc_start = fc.index.min()

        # --- Observations before forecast ---
        obs_crop = obs.loc[:fc_start - pd.Timedelta(hours=1)]

        # --- Combine forcing ---
        merged = pd.concat([obs_crop, fc])

        # --- Add observed discharge everywhere ---
        merged["q"] = 1 #obs["q"]

        # --- Keep correct order ---
        cols = (
            [f"rr{i}" for i in range(10)] +
            [f"tm{i}" for i in range(10)] +
            ["q"]
        )

        merged = merged[cols]

        # --- Cut at forecast end (optional but you said you want this) ---
        fc_end = fc.index.max()
        merged = merged.loc[:fc_end]

        # --- Save ---
        merged.reset_index().to_csv(file, index=False)

        print(f"✔ Merged: {os.path.basename(file)}")

    except Exception as e:
        print(f"❌ Error in {file}: {e}")

print("All files merged correctly ✅")
```
