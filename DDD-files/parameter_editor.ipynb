## Parameter file editor

Function to calculate the Gamma scale parameter for parameter files generated through HSO, and add this to the parameter file. Input is file path to parameter file, and name of catchment. 

$G_{sh}=1$

$G_{sc}=\frac{\Delta t}{MRT}$

```python
import pandas as pd
```

```python
# catchment = "Gaupne"
# catchment = "Austdalsvatnet"
# catchment = "Tunsbergdalsvatnet"
# catchment= "Myklemyr"
# catchment = "BåtedølaS"
# catchment = "Geisdalsvatnet"
# catchment = "Fonndøla"
catchment = "Nigard"
```

```python
def parameter_editor(file_path, catchment):
    df = pd.read_csv(file_path, sep=";", header=None)

    print(f"{catchment}:")
    print(f"MRT = {df.iloc[31,1]} days.")
    print(f"Timeresinsec = {df.iloc[26,1]} seconds.")

    GshInt = 1
    GscInt = float(df.iloc[26,1]) / (float(df.iloc[31,1]) * 3600 * 24)
    print("GscInt =", GscInt)

    df.loc[31, 0] = "GshInt"
    df.loc[31, 1] = GshInt

    new_row = pd.DataFrame([["GscInt", GscInt]])
    df = pd.concat([df.iloc[:32], new_row, df.iloc[32:]]).reset_index(drop=True)
    
    new_file_path = file_path.replace(".csv", "_edited.csv")
    df.to_csv(new_file_path, sep=";", header=False, index=False)

    print(f"File saved as: {new_file_path}")
    return df

file_path = f"{catchment}_1H_param.csv"
# file_path = f"{catchment}_3H_param.csv"
parameter_editor(file_path, catchment)
```

### For fixing the HSO bug with glacier fraction: 


```python
import pandas as pd
import numpy as np

# catchment = "Gaupne"
# catchment = "Austdalsvatnet"
catchment = "Tunsbergdalsvatnet"
# catchment = "Myklemyr"
# catchment = "BåtedølaS"
# catchment = "Geisdalsvatnet"

def parameter_editor_2(file_path, catchment):
    df = pd.read_csv(file_path, sep=";", header=None)

    print(f"{catchment}:")
    print(f"Glacier fraction = {df.iloc[41,1]*100} %")
    print(f"Catchment area = {df.iloc[28, 1]} m^2")
    print(" ")
    print("Previous values")
    print(f"maxGl = {df.iloc[45,1]} ")
    print(f"midGl = {df.iloc[50,1]} ")
    print(f"stdGl = {df.iloc[52,1]} ")
    
    glac_frac = df.iloc[41,1]
    catchment_area = df.iloc[28, 1]
    glac_area=glac_frac*catchment_area
    
    
    maxGl_new = round((4*glac_area/np.pi)**(1/2), 2)
    midGl_new = round(maxGl_new/2, 2)
    stdGl_new = round(midGl_new/2, 2)
    
    df.iloc[45, 1] = maxGl_new
    df.iloc[50, 1] = midGl_new
    df.iloc[52, 1] = stdGl_new
    print(" ")     
    print("New values")
    print(f"maxGl_new = {df.iloc[45,1]} ")
    print(f"midGl_new = {df.iloc[50,1]} ")
    print(f"stdGl_new = {df.iloc[52,1]} ")
    


    df.to_csv(file_path, sep=";", header=False, index=False)

    print(f"File saved as: {file_path}")
    return df

file_path = f"{catchment}_3H_param_edited.csv"
parameter_editor_2(file_path, catchment)
```
